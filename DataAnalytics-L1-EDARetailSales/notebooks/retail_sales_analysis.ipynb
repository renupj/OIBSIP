```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
category_sales
```

```python
import matplotlib.pyplot as plt

category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
region_sales
```

```python
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
profit_category
```

```python
profit_category.plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.head()
```

```python
monthly_sales.plot(kind="line", figsize=(12, 5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
top_products.sort_values().plot(kind="barh", figsize=(10, 6))
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
top_profit_products = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_profit_products
```

```python
top_profit_products.sort_values().plot(kind="barh", figsize=(10, 6))
plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

segment_sales
```

```python
segment_sales.plot(kind="bar")
plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
profit_region = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

profit_region
```

```python
profit_region.plot(kind="bar")
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
id="9j2k7p"
discount_profit = df.groupby("Discount")["Profit"].mean()
discount_profit
```

```python
id="c4m8rx"
discount_profit.plot(kind="line", marker="o")
plt.title("Average Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.grid(True)
plt.show()
```
